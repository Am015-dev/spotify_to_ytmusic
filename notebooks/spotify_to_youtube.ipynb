# Spotify to YouTube Music Migration
This notebook demonstrates how to migrate your Spotify playlists and liked songs to YouTube Music using the **spotify2ytmusic** tool. Each cell contains commands you can run in a terminal-enabled environment.

## 1. Create and Activate an Environment
We'll use conda to manage the environment for this example.

```bash
!conda create --name spotify_to_youtube -y
!source "$(conda info --base)/etc/profile.d/conda.sh"
!conda activate spotify_to_youtube
```

## 2. Install the Tool
Install Python (if it's not already present) and the `spotify2ytmusic` package.

```bash
!pip install python
!pip install spotify2ytmusic
```

## 3. Prepare a Working Directory
Create a folder to store your Spotify export and YouTube Music credentials.

```bash
!mkdir -p spotify_to_youtube
%cd spotify_to_youtube
```

## 4. Authenticate with YouTube Music
This step opens a browser window for OAuth. Follow the prompts to create the authentication file in the current directory.

```bash
!ytmusicapi oauth
```

## 5. Export Spotify Playlists
Download the `spotify-backup.py` script and export your playlists and liked songs.

```bash
!curl -L -O https://raw.githubusercontent.com/caseychu/spotify-backup/master/spotify-backup.py
!python spotify-backup.py playlists.json --dump=liked,playlists --format=json
```

## 6. Import Liked Songs into YouTube Music
Use the `spotify2ytmusic` CLI to like the exported tracks on YouTube Music.

```bash
!s2yt_load_liked
```

That's it! You can now enjoy your Spotify music collection on YouTube Music.
